```python
import pandas as pd
import sklearn as sk
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#step1 handle data 

# train.csv
#a)       → List of (stock_id, time_id) + target (future 10-min realized vol) (target(realized volaoty) we wantto score on)

```

```python
train = pd.read_csv("data/train.csv")
train.head(3)
train.info()
#train-> time_id time of the "id" for 10mnt and we have it target 
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(9,2))
plt.hist(train["target"], bins=train["stock_id"].nunique())
plt.title("Target Distribution")
plt.xlabel("vol")
plt.ylabel("Frequcy")
```

```python
print("25%:", train["target"].quantile(0.25))
print("50%:", train["target"].quantile(0.50))
print("75%:", train["target"].quantile(0.75))
```

```python
stock_mean = train.groupby("stock_id")["target"].mean()
plt.figure(figsize=(9,2))
plt.scatter(stock_mean.index, stock_mean.values)
plt.title("Mean vol / stck")
plt.xlabel("stock ID")
plt.ylabel("Mean targt")
plt.show()
plt.figure(figsize=(9,2))
plt.xlabel("diff 10mnt windows ")
plt.plot(train.groupby("time_id")["target"].mean().values)
plt.ylabel("Mean Target")
plt.show()
```
